# Kontrollstrukturen in Python

Kontrollstrukturen steuern den Ablauf eines Programms: in welcher Reihenfolge Anweisungen ausgeführt werden, welche Anweisungen nur unter bestimmten Bedingungen laufen und welche wiederholt werden.

Ein Kochrezept funktioniert genauso: Die Schritte werden der Reihe nach abgearbeitet ("Mehl abwiegen, Eier dazugeben"), manche nur unter einer Bedingung ("**wenn** der Teig zu fest ist, Milch dazugeben"), und manche werden wiederholt ("rühren, **bis** der Teig glatt ist").

![Arten von Kontrollstrukturen](flowchart-overview-text-3-done.png)

*Arten von Kontrollstrukturen. Quelle: [Core Electronics](https://core-electronics.com.au/guides/control-structure-python/)*

Die Grafik zeigt die drei grundlegenden Konzepte:

- **Sequenz**: Alle Anweisungen werden nacheinander abgearbeitet, genau so, wie sie im Code stehen. Das ist der Standardfall.
- **Selektion**: Der Programmfluss verzweigt sich anhand einer Bedingung. Je nachdem, ob sie erfüllt ist, wird ein anderer Pfad ausgeführt.
- **Iteration**: Ein oder mehrere Schritte werden wiederholt, solange eine Bedingung erfüllt ist oder noch Elemente einer Datenstruktur übrig sind.

⚠️ Zwei Beispielzellen im Kapitel zur `while`-Schleife verhalten sich besonders: Eine wartet auf eine Eingabe, die andere ist eine Endlosschleife und deshalb auskommentiert. Beide sind im Text markiert.

## 1. Bedingte Anweisungen

Bedingte Anweisungen führen Code nur dann aus, wenn eine bestimmte Bedingung erfüllt ist. Damit lassen sich Verzweigungen in den Programmablauf einbauen.

<a id="if-anweisung"></a>

### 1.1 Die `if`-Anweisung

Mit dem Schlüsselwort `if` wird ein Teil des Codes nur dann ausgeführt, wenn eine Bedingung erfüllt ist. Eine Bedingung ist ein Ausdruck, der als Boolean ausgewertet wird, also `True` oder `False` ergibt.

![If-statement](if-statement.jpg)

*If-Anweisung Flowchart. Quelle: [GeeksforGeeks](https://www.geeksforgeeks.org/python/python3-if-if-else-nested-if-if-elif-statements/)*

Die Schreibweise besteht aus drei Teilen:

```python
if bedingung:
    anweisung_1
    anweisung_2
weiter_nach_dem_if
```

- Hinter der Bedingung steht ein **Doppelpunkt**.
- Alle Zeilen, die zum `if` gehören, sind **eingerückt**, üblicherweise mit 4 Leerzeichen. Die Einrückung bestimmt in Python, welche Zeilen einen Block bilden.
- Die erste Zeile, die wieder weiter links steht, gehört nicht mehr zum Block und wird in jedem Fall ausgeführt.

````python
if True:
    print("Wird ausgeführt, weil die Bedingung True ist.")

if False:
    print("Wird NICHT ausgeführt, weil die Bedingung False ist.")

print("Steht nicht mehr im Block und wird immer ausgeführt.")
````

In der Praxis steht statt `True` oder `False` ein Vergleich. Alle Vergleichs- und logischen Operatoren aus der Session "Datentypen" lassen sich verwenden:

````python
temperature = 28

if temperature > 25:
    print("Es ist warm.")

if temperature >= 20 and temperature <= 30:
    print("Die Temperatur liegt zwischen 20 und 30 Grad.")

if temperature < 0 or temperature > 35:
    print("Extreme Temperatur (wird hier NICHT ausgeführt).")

if not temperature == 0:
    print("Die Temperatur ist nicht genau 0 Grad.")
````

Mit `in` lässt sich prüfen, ob ein Element in einer Datenstruktur enthalten ist:

````python
weekend_days = ["Samstag", "Sonntag"]
today = "Samstag"

if today in weekend_days:
    print(f"{today} ist ein Wochenendtag.")

if "Montag" in weekend_days:
    print("Montag ist ein Wochenendtag (wird NICHT ausgeführt).")
````

Aus der Session "Datentypen" ist bekannt, dass jeder Wert als Boolean ausgewertet werden kann. `0`, `""`, `None` und leere Datenstrukturen gelten als `False`, alle anderen Werte als `True`. Das wird in Bedingungen oft genutzt, um zu prüfen, ob überhaupt etwas vorhanden ist:

````python
shopping_list = ["Milch"]
note = ""

if shopping_list:
    print("Die Einkaufsliste enthält mindestens einen Eintrag.")

if note:
    print("Es gibt eine Notiz (wird NICHT ausgeführt, leerer String).")

if "False":
    print('"False" ist ein nicht leerer String und gilt deshalb als True.')
````

Mehrere Bedingungen lassen sich in einer einzigen `if`-Anweisung kombinieren:

````python
age = 34
has_ticket = True
banned_guests = ["Max", "Erika"]
name = "Anna"

if age >= 18 and has_ticket and name not in banned_guests:
    print(f"{name} darf eintreten.")
````

Dieselbe Logik lässt sich auch mit verschachtelten `if`-Anweisungen schreiben. Das ist möglich, aber schwerer zu lesen:

````python
if age >= 18:
    if has_ticket:
        if name not in banned_guests:
            print(f"{name} darf eintreten.")
````

### 1.3 Der `else`-Block

Mit `else` nach einem `if`-Block wird eine Alternative festgelegt, die ausgeführt wird, wenn die Bedingung `False` ist. Von beiden Blöcken wird also immer genau einer ausgeführt.

![If-statement](if-else.jpg)

<a id="elif-anweisung"></a>

### 1.4 Die `elif`-Anweisung

Mit `elif` (kurz für *else if*) werden mehrere Bedingungen nacheinander geprüft. Python prüft von oben nach unten und führt nur den **ersten** Block aus, dessen Bedingung zutrifft. Alle weiteren werden übersprungen. Trifft keine Bedingung zu, läuft der `else`-Block.

![If-statement](if-elseif-ladder.jpg)

*If-Elif-Else Flowchart. Quelle: [GeeksforGeeks](https://www.geeksforgeeks.org/python/python3-if-if-else-nested-if-if-elif-statements/)*

````python
points = 78

if points >= 90:
    grade = "sehr gut"
elif points >= 75:
    grade = "gut"
elif points >= 60:
    grade = "befriedigend"
elif points >= 50:
    grade = "ausreichend"
else:
    grade = "mangelhaft"

print(f"Note: {grade}")
````

⚠️ Die Reihenfolge der Bedingungen ist entscheidend: Bei `78` Punkten trifft auch `points >= 60` zu. Da aber `points >= 75` vorher geprüft wird und zutrifft, werden die weiteren Zweige nicht mehr betrachtet. Stünde `points >= 50` an erster Stelle, wäre jedes Ergebnis ab 50 Punkten "ausreichend".

## 2. Schleifen

Schleifen führen eine Folge von Anweisungen wiederholt aus. Python kennt zwei Arten: Die `while`-Schleife wiederholt, **solange** eine Bedingung erfüllt ist. Die `for`-Schleife wiederholt **für jedes** Element einer Datenstruktur.

<a id="while-schleife"></a>

### 2.1 Die `while`-Schleife

Die `while`-Schleife ist aufgebaut wie eine `if`-Anweisung. Der Unterschied: Nach dem Block wird die Bedingung erneut geprüft. Ist sie weiterhin `True`, läuft der Block noch einmal, bis die Bedingung `False` ist.

```python
while bedingung:
    anweisungen   # werden wiederholt, solange die Bedingung True ist
```

![While-Schleife](while-loop.jpg)

*While-Schleife Flowchart. Quelle: [GeeksforGeeks](https://www.geeksforgeeks.org/python/python-while-loop/)*

In Schleifen wird häufig eine Variable um einen Wert erhöht oder verringert. Dafür gibt es eine Kurzschreibweise: `count -= 1` bedeutet dasselbe wie `count = count - 1`, `count += 1` dasselbe wie `count = count + 1`.

````python
# Countdown von 5 auf 1
count = 5

while count > 0:
    print(f"Aktueller Zählerstand: {count}")
    count -= 1

print("Start!")
print(f"Wert von count nach der Schleife: {count}")
````

#### `break` und `continue`

Schleifen lassen sich mit zwei Anweisungen steuern:

- `break` beendet die Schleife sofort und vollständig.
- `continue` bricht nur den aktuellen Durchlauf ab und springt zum nächsten.

````python
# break: Die Schleife läuft eigentlich endlos, bis break sie beendet
number = 0

while True:
    print(number)
    number += 1
    if number > 5:
        break

print("Schleife wurde mit break beendet.")
````

Dasselbe Ergebnis lässt sich hier auch ohne `break` erreichen, indem die Abbruchbedingung direkt in die `while`-Bedingung geschrieben wird. Das ist meist übersichtlicher. `break` lohnt sich vor allem, wenn erst mitten im Block feststeht, ob abgebrochen werden soll.

````python
number = 0

while number <= 5:
    print(number)
    number += 1
````

````python
# continue: gerade Zahlen überspringen
number = 0

while number < 10:
    number += 1
    if number % 2 == 0:
        continue
    print(number)

print("Es wurden nur ungerade Zahlen ausgegeben.")
````

#### Endlosschleifen

Eine `while`-Schleife läuft endlos, wenn sich im Block nichts ändert, was die Bedingung irgendwann `False` macht, und kein `break` erreicht wird.

⚠️ Die folgende Zelle ist eine **unbeabsichtigte Endlosschleife**, da `number` nie verändert wird. Sie ist deshalb auskommentiert. Zum Ausprobieren die Kommentarzeichen vor den beiden letzten Zeilen entfernen und die laufende Zelle anschließend mit dem Stopp-Button (■, *Interrupt*) in der Notebook-Leiste abbrechen.

````python
number = 5

# while number > 0:
#     print("Dies ist eine Endlosschleife.")
````

Eine **beabsichtigte** Endlosschleife mit `while True` ist dagegen üblich, wenn die Schleife so lange laufen soll, bis eine bestimmte Eingabe kommt. Die Funktion `input()` wartet dabei auf eine Eingabe über die Tastatur.

⚠️ Die folgende Zelle wartet auf Eingaben und läuft, bis `exit` eingegeben wird.

````python
while True:
    user_input = input("Gib 'exit' ein, um zu beenden: ")
    if user_input == "exit":
        print("Die while-Schleife wurde beendet.")
        break
    elif user_input == "hallo":
        print("Hallo zurück!")
    else:
        print(f"Eingabe: '{user_input}'")
````

#### `while` und Datenstrukturen

Mit einem Zähler als Index lässt sich eine Liste auch mit `while` durchlaufen:

````python
participants = ["Anna", "Bernd", "Claudia", "Thomas"]

index = 0
while index < len(participants):
    print(f"Teilnehmer:in: {participants[index]}")
    index += 1
````

Das funktioniert, ist aber umständlich: Index anlegen, Bedingung formulieren, Index erhöhen. Vergisst man die letzte Zeile, entsteht eine Endlosschleife. Für genau diesen Fall gibt es die `for`-Schleife.

<a id="for-schleife"></a>

### 2.2 Die `for`-Schleife

Die `for`-Schleife durchläuft nacheinander **jedes Element** einer Datenstruktur. In jedem Durchlauf steht das aktuelle Element in einer Variable zur Verfügung. Vergleichbar ist das mit einem Stapel Briefe, den man von oben nach unten abarbeitet: Jeder Brief wird einmal in die Hand genommen, bis der Stapel leer ist.

```python
for element in datenstruktur:
    anweisungen   # werden für jedes Element einmal ausgeführt
```

In anderen Programmiersprachen heißt diese Art Schleife oft *for-each*.

![For-Schleife](for-loop-python.jpg)

*For-Schleife Flowchart. Quelle: [GeeksforGeeks](https://www.geeksforgeeks.org/python/python-for-loops/)*

Das `while`-Beispiel von oben mit `for`:

````python
participants = ["Anna", "Bernd", "Claudia", "Thomas"]

for participant in participants:
    print(f"Teilnehmer:in: {participant}")
````

Die `for`-Schleife funktioniert mit allen Datenstrukturen aus der Session "Datenstrukturen" und auch mit Strings. Bei Sets ist die Reihenfolge nicht festgelegt, bei Dictionaries werden die **Schlüssel** durchlaufen:

````python
for letter in "Heureka":
    print(letter)
````

````python
primes = (2, 3, 5, 7)
for prime in primes:
    print(prime)
````

````python
prices = {"Kaffee": 4.99, "Brot": 2.49, "Butter": 1.99}
for product in prices:
    print(f"{product} kostet {prices[product]} Euro.")
````

In jedem Durchlauf kann beliebiger Code stehen, auch Berechnungen und `if`-Anweisungen. `break` und `continue` funktionieren genauso wie bei `while`:

````python
numbers = [1, 2, 3, 4, 5]

for number in numbers:
    square = number ** 2
    print(f"Das Quadrat von {number} ist {square}")
````

````python
# break: Suche beenden, sobald die 3 gefunden ist
for number in numbers:
    if number == 3:
        print("Zahl 3 gefunden, Schleife wird beendet.")
        break
    print(f"Zahl: {number}")
````

````python
# continue: gerade Zahlen überspringen
for number in numbers:
    if number % 2 == 0:
        continue
    print(f"Ungerade Zahl: {number}")
````

Häufig wird in einer Schleife ein Ergebnis schrittweise aufgebaut: Vor der Schleife wird eine Variable angelegt, in jedem Durchlauf wird sie ergänzt.

````python
prices_list = [4.99, 2.49, 1.99, 6.79]

total = 0
cheap_prices = []

for price in prices_list:
    total += price
    if price < 3:
        cheap_prices.append(price)

print(f"Summe: {round(total, 2)}")
print(f"Preise unter 3 Euro: {cheap_prices}")
````

#### Zahlenfolgen mit `range()`

Soll eine Schleife eine bestimmte Anzahl von Durchläufen haben, erzeugt `range()` die passende Zahlenfolge:

- `range(5)`: die Zahlen `0` bis `4`
- `range(2, 6)`: die Zahlen `2` bis `5` (das Ende ist wie beim Slicing exklusiv)
- `range(0, 20, 3)`: von `0` bis unter `20` in Dreierschritten

````python
for i in range(5):
    print(i)
````

````python
for i in range(0, 20, 3):
    print(i)
````

💡 Der Variablenname `i` (für *Index*) ist bei Zählschleifen mit `range()` eine verbreitete Konvention. Sonst sind sprechende Namen wie `participant` oder `price` vorzuziehen.

#### Index und Element mit `enumerate()`

Wird neben dem Element auch seine Position gebraucht, liefert `enumerate()` in jedem Durchlauf ein Paar aus Index und Element. Mit zwei Variablennamen hinter `for` wird das Paar direkt aufgeteilt:

````python
participants = ["Anna", "Bernd", "Claudia"]

for index, participant in enumerate(participants):
    print(f"{index}: {participant}")
````

Mit `enumerate(..., start=1)` beginnt die Zählung bei 1, etwa für eine nummerierte Liste:

````python
for position, participant in enumerate(participants, start=1):
    print(f"{position}. {participant}")
````

#### Schlüssel und Wert mit `.items()`

Für Dictionaries funktioniert das genauso: `.items()` liefert Paare aus Schlüssel und Wert, die mit zwei Variablennamen aufgeteilt werden:

````python
scores = {"Anna": 85, "Bernd": 78, "Claudia": 92}

for name, score in scores.items():
    print(f"{name} hat {score} Punkte erzielt.")
````

#### Verschachtelte Schleifen

Schleifen lassen sich ineinander verschachteln. Die innere Schleife läuft dabei bei **jedem** Durchlauf der äußeren komplett durch.

Der Parameter `end=" "` von `print()` sorgt dafür, dass nach der Ausgabe ein Leerzeichen statt eines Zeilenumbruchs folgt. Ein leeres `print()` beendet dann die Zeile.

````python
for participant in participants:
    for letter in participant:
        print(letter, end=" ")
    print()  # neue Zeile nach jedem Namen
````

## 3. Zusammenfassung

Am Anfang standen drei Konzepte: **Sequenz**, **Selektion** und **Iteration**. Die Sequenz ist der Standardfall, Anweisungen werden von oben nach unten abgearbeitet. Für Selektion und Iteration stellt Python eigene Konstrukte bereit:

| Konzept | Werkzeug in Python |
|---|---|
| Sequenz | Anweisungen von oben nach unten |
| Selektion | `if` / `elif` / `else` |
| Iteration | `while` / `for` |

### Behandelte Konzepte

1. Mit `if`, `elif` und `else` steuern, welcher Code ausgeführt wird, und dabei Doppelpunkt und Einrückung beachten.
2. Bedingungen formulieren und mit `and`, `or`, `not` und `in` kombinieren.
3. Mit einer `while`-Schleife wiederholen, solange eine Bedingung gilt.
4. Mit einer `for`-Schleife jedes Element einer Datenstruktur durchlaufen und mit `range()`, `enumerate()` und `.items()` kombinieren.
5. Schleifen mit `break` und `continue` steuern.
6. Einschätzen, wann `while` und wann `for` das passende Werkzeug ist.

### Wahl der passenden Schleife

| Situation | Werkzeug |
|---|---|
| Eine Sammlung (Liste, String, Dictionary, ...) durchlaufen | `for` |
| Anzahl der Durchläufe steht vorher fest | `for` mit `range(...)` |
| Abbruch hängt an einer Bedingung, Anzahl unbekannt | `while` |
| Mehrere Fälle, die sich gegenseitig ausschließen | eine `elif`-Kette statt mehrerer `if` |
| Index *und* Element gleichzeitig benötigt | `for ... in enumerate(...)` |
| Schlüssel *und* Wert eines Dictionaries benötigt | `for ... in ....items()` |

### Kurzreferenz

| Schreibweise | Bedeutung |
|---|---|
| `if condition:` | Block nur ausführen, wenn die Bedingung wahr ist |
| `elif condition:` | weitere Bedingung prüfen, falls die vorherigen nicht zutrafen |
| `else:` | Block ausführen, wenn keine Bedingung zutraf |
| `pass` | Platzhalter für einen leeren Block |
| `while condition:` | Block wiederholen, solange die Bedingung wahr ist |
| `for item in collection:` | Block für jedes Element ausführen |
| `range(start, end, step)` | Zahlenfolge, Ende exklusiv |
| `for index, item in enumerate(collection):` | Index und Element durchlaufen |
| `for key, value in my_dict.items():` | Schlüssel und Wert durchlaufen |
| `break` | Schleife sofort beenden |
| `continue` | zum nächsten Durchlauf springen |
| `count += 1` | Kurzform für `count = count + 1` |

### Typische Fehlerquellen

- **Endlosschleife**: In der `while`-Schleife muss sich etwas ändern, das die Abbruchbedingung irgendwann falsch macht (z.B. `count -= 1`), oder es braucht ein `break`.
- **`=` statt `==`**: `=` weist zu, `==` vergleicht. In einer Bedingung ist fast immer `==` gemeint.
- **Einrückung**: Sie entscheidet über die Blockzugehörigkeit. Eine Zeile zu weit links wird nach der Schleife ausgeführt statt in ihr.
- **Reihenfolge in `elif`-Ketten**: Nur der erste zutreffende Zweig wird ausgeführt. Allgemeinere Bedingungen gehören deshalb nach unten.

### 4.1 Der `else`-Zweig in Schleifen

Schleifen können einen `else`-Zweig haben. Er wird ausgeführt, wenn die Schleife **regulär** endet, und **übersprungen**, wenn sie mit `break` verlassen wurde. Das eignet sich für Suchen: Wird das Gesuchte gefunden, bricht `break` ab, sonst meldet der `else`-Zweig, dass nichts gefunden wurde.

````python
participants = ["Anna", "Bernd", "Claudia"]
searched_name = "Dora"

for participant in participants:
    if participant == searched_name:
        print(f"{searched_name} gefunden.")
        break
else:
    print(f"{searched_name} ist nicht in der Liste.")
````

Wird `searched_name` auf `"Bernd"` gesetzt, endet die Schleife mit `break` und der `else`-Zweig läuft nicht. Dasselbe funktioniert bei `while`-Schleifen.

<a id="iteratoren"></a>

### 4.2 Iteratoren

Unter der Haube arbeitet die `for`-Schleife mit einem **Iterator**. Ein Iterator ist ein Objekt, das auf einer Datenstruktur aufsetzt und sich merkt, welches Element als nächstes an der Reihe ist. Mit `next()` wird das nächste Element abgerufen.

![Iterator](iterator.png)

*Quelle: [Machine Learning Geek](https://machinelearninggeek.com/python-iterators-examples/)*

````python
numbers = [1, 2, 3]
numbers_iterator = iter(numbers)
numbers_iterator
````

````python
# Diese Zelle mehrmals ausführen und auf die jeweilige Ausgabe achten
next(numbers_iterator)
````

Nach dem dritten Aufruf ist der Iterator erschöpft, ein weiterer Aufruf löst den Fehler `StopIteration` aus.

<a id="exception-handling"></a>

### 4.3 Iteratoren und Exception Handling

Fehler wie `StopIteration` lassen sich mit einer weiteren Kontrollstruktur abfangen: `try` und `except` (ausführlich in der Session "Fehlerbehandlung"):

````python
# Ein Iterator mit zwei Elementen wird dreimal abgefragt
short_iterator = iter(["a", "b"])

for attempt in range(3):
    try:
        print(next(short_iterator))
    except StopIteration:
        print("Keine weiteren Elemente im Iterator.")
````

Damit lässt sich nachbauen, was eine `for`-Schleife intern tut:

````python
even_numbers = {2, 4, 6, 8, 10}
even_iterator = iter(even_numbers)

while True:
    try:
        element = next(even_iterator)
    except StopIteration:
        print("Alle Elemente wurden verarbeitet.")
        break
    print(element)
````
