# Erweiterte Python-Syntax (Syntactic Sugar)

Python bietet für viele häufige Aufgaben eine kürzere Schreibweise, als man sie mit den bisher bekannten Mitteln bräuchte. Man spricht von **Syntactic Sugar**: Die Schreibweisen ermöglichen nichts grundsätzlich Neues, machen Code aber kürzer und oft lesbarer. Vergleichbar ist das mit Abkürzungen im Alltag: "z.B." statt "zum Beispiel" spart Platz, solange alle Beteiligten die Abkürzung kennen.

Genau darin liegt auch die Grenze: Kurzschreibweisen helfen nur, solange der Code dadurch verständlicher wird. Einige der folgenden Konstrukte lassen sich so weit treiben, dass der Code schwerer zu lesen ist als die ausführliche Variante.

⚠️ Eine Beispielzelle bricht **absichtlich** mit einem Fehler ab, sie ist im Text markiert. Einige Zellen legen Textdateien in diesem Ordner an.

## 1. Ternärer Operator

Für eine einfache Entscheidung zwischen zwei Werten braucht eine `if`-`else`-Anweisung vier Zeilen:

````python
first_value = 5
second_value = 10

if first_value > second_value:
    maximum = first_value
else:
    maximum = second_value

print(f"Maximum: {maximum}")
````

Der **ternäre Operator** (*conditional expression*) fasst das in einem einzigen Ausdruck zusammen:

```python
value_if_true if condition else value_if_false
```

Man liest ihn wie einen Satz: "`first_value`, wenn `first_value > second_value`, sonst `second_value`".

````python
maximum = first_value if first_value > second_value else second_value
print(f"Maximum: {maximum}")
````

Weil der ternäre Operator ein Ausdruck ist, lässt er sich überall einsetzen, wo ein Wert erwartet wird, zum Beispiel direkt in einem f-String:

````python
age = 19
print(f"Status: {'volljährig' if age >= 18 else 'minderjährig'}")
````

⚠️ Ternäre Operatoren lassen sich verschachteln, das wird aber schnell unübersichtlich. Ab mehr als zwei Fällen ist eine `if`-`elif`-`else`-Kette die lesbarere Wahl:

````python
age = 70

# Möglich, aber schwer lesbar
status = "Kind" if age < 18 else "Senior" if age >= 65 else "Erwachsen"
print(f"Status: {status}")

# Besser lesbar
if age < 18:
    status = "Kind"
elif age >= 65:
    status = "Senior"
else:
    status = "Erwachsen"
print(f"Status: {status}")
````

## 2. Comprehensions

Ein häufiges Muster aus der Session "Kontrollstrukturen": Eine leere Liste anlegen, mit einer `for`-Schleife durchlaufen und in jedem Durchlauf ein Element anhängen. **Comprehensions** schreiben genau dieses Muster in einer einzigen Zeile.

<a id="list-comprehensions"></a>

### 2.1 List Comprehensions

````python
# Mit Schleife
squares = []
for number in range(10):
    squares.append(number**2)
print(f"Mit Schleife: {squares}")
````

````python
# Mit List Comprehension
squares = [number**2 for number in range(10)]
print(f"Mit List Comprehension: {squares}")
````

Der Aufbau entspricht der Schleife, nur in anderer Reihenfolge: Vorne steht, **was** in die neue Liste kommt, dahinter, **woher** die Werte stammen.

```python
[expression for item in collection]
```

#### Filtern mit `if`

Mit einem `if` am Ende werden nur die Elemente übernommen, für die die Bedingung zutrifft:

````python
even_squares = [number**2 for number in range(10) if number % 2 == 0]
print(f"Quadrate der geraden Zahlen: {even_squares}")
````

#### Umformen mit dem ternären Operator

Soll dagegen **jedes** Element übernommen, aber je nach Bedingung unterschiedlich umgeformt werden, steht ein ternärer Operator **vorne** im Ausdruck:

````python
labels = [number**2 if number % 2 == 0 else "ungerade" for number in range(10)]
print(labels)
````

⚠️ Die Position entscheidet über die Bedeutung:

- `[x for x in items if condition]`: `if` **hinten**, ohne `else` → **filtert** Elemente heraus
- `[a if condition else b for x in items]`: `if`-`else` **vorne** → **formt** jedes Element um, die Anzahl bleibt gleich

#### Comprehensions statt `map()` und `filter()`

Aus der Session "Funktionen" sind `map()` und `filter()` mit Lambda-Funktionen bekannt. Comprehensions erledigen dieselben Aufgaben und gelten in Python meist als lesbarer:

````python
numbers = [1, 2, 3, 4, 5]

# map() und filter() mit Lambda
doubled = list(map(lambda number: number * 2, numbers))
evens = list(filter(lambda number: number % 2 == 0, numbers))

# Dasselbe mit Comprehensions
doubled = [number * 2 for number in numbers]
evens = [number for number in numbers if number % 2 == 0]

print(f"{doubled = }")
print(f"{evens = }")
````

Ein praxisnahes Beispiel: Rohdaten bereinigen.

````python
raw_names = ["  anna ", "BERND", " claudia", ""]

clean_names = [name.strip().title() for name in raw_names if name.strip()]
print(clean_names)
````

<a id="dict-comprehensions"></a>

### 2.2 Dictionary Comprehensions

Dictionary Comprehensions funktionieren genauso, mit geschweiften Klammern und einem Paar `key: value` vorne:

````python
squares_by_number = {number: number**2 for number in range(6)}
print(squares_by_number)
````

````python
prices = {"Kaffee": 4.99, "Brot": 2.49, "Käse": 6.79, "Butter": 1.99}

# Nur Produkte unter 3 Euro
cheap_products = {product: price for product, price in prices.items() if price < 3}
print(cheap_products)

# Alle Preise um 10 % erhöhen
increased_prices = {product: round(price * 1.1, 2) for product, price in prices.items()}
print(increased_prices)
````

<a id="set-comprehensions"></a>

### 2.3 Set Comprehensions

Mit geschweiften Klammern und einem einzelnen Ausdruck vorne entsteht ein Set. Doppelte Werte fallen dabei wie gewohnt weg:

````python
emails = ["anna@firma.de", "bernd@beispiel.com", "claudia@firma.de"]

domains = {email.split("@")[1] for email in emails}
print(domains)
````

💡 Comprehensions lohnen sich, solange sie in eine gut lesbare Zeile passen. Sobald mehrere Bedingungen, verschachtelte Schleifen oder Nebeneffekte wie `print()` dazukommen, ist eine normale `for`-Schleife meist die bessere Wahl.

## 3. Tuple Unpacking und der `*`-Operator

Beim **Tuple Unpacking** werden die Elemente einer Sequenz in einem Schritt auf mehrere Variablen verteilt. Es funktioniert mit Tupeln, Listen und allen anderen Sequenzen.

````python
location = (51.34, 12.37)

# Einzeln über den Index
latitude = location[0]
longitude = location[1]

# Mit Unpacking
latitude, longitude = location
print(f"{latitude = }, {longitude = }")
````

Unpacking ist bereits aus der Session "Kontrollstrukturen" bekannt: `for index, participant in enumerate(...)` und `for key, value in my_dict.items()` verteilen in jedem Durchlauf ein Paar auf zwei Variablen.

Ein typischer Anwendungsfall ist das Vertauschen zweier Variablen ohne Hilfsvariable:

````python
first_value, second_value = 10, 20
print(f"Vorher: {first_value = }, {second_value = }")

first_value, second_value = second_value, first_value
print(f"Nachher: {first_value = }, {second_value = }")
````

⚠️ Die Anzahl der Variablen muss zur Anzahl der Elemente passen. Die folgende Zelle bricht **absichtlich** mit einem `ValueError` ab:

````python
first_value, second_value = [1, 2, 3]
````

### Restliche Elemente mit `*` sammeln

Mit einem `*` vor einem Variablennamen sammelt diese Variable alle übrigen Elemente als Liste:

````python
ranking = ["Claudia", "Anna", "Bernd", "David", "Eve"]

winner, *others = ranking
print(f"{winner = }, {others = }")

first, *middle, last = ranking
print(f"{first = }, {middle = }, {last = }")
````

### Sequenzen und Dictionaries beim Aufruf entpacken

Aus der Session "Funktionen" ist bekannt, dass `*args` und `**kwargs` beliebig viele Argumente **entgegennehmen**. Umgekehrt **verteilen** `*` und `**` beim Aufruf einer Funktion eine Liste bzw. ein Dictionary auf einzelne Argumente:

````python
def describe_trip(origin, destination, distance_km):
    """Return a short description of a trip."""
    return f"{origin} → {destination}: {distance_km} km"


trip_as_list = ["Leipzig", "Berlin", 190]
trip_as_dict = {"origin": "Köln", "destination": "Bonn", "distance_km": 30}

print(describe_trip(*trip_as_list))    # wie describe_trip("Leipzig", "Berlin", 190)
print(describe_trip(**trip_as_dict))   # wie describe_trip(origin="Köln", ...)
````

Mit `**` lassen sich auch zwei Dictionaries zu einem neuen zusammenführen. Bei gleichen Schlüsseln gewinnt der Wert aus dem hinteren Dictionary:

````python
default_settings = {"theme": "light", "font_size": 12}
user_settings = {"font_size": 14}

settings = {**default_settings, **user_settings}
print(settings)
````

## 4. f-Strings formatieren

f-Strings sind seit der Session "Datentypen" bekannt, ebenso die Kurzform `f"{expression = }"`. Zusätzlich lässt sich hinter einem Doppelpunkt angeben, **wie** ein Wert dargestellt werden soll:

```python
f"{value:format_spec}"
```

````python
pi = 3.14159265359

print(f"{pi}")        # unformatiert
print(f"{pi:.2f}")    # 2 Nachkommastellen
print(f"{pi:.2e}")    # wissenschaftliche Notation
print(f"{0.1234:.1%}")  # als Prozentwert
````

````python
amount = 1234567.897654

print(f"{amount:,.2f}")   # Tausendertrennzeichen und 2 Nachkommastellen
print(f"{amount:_.2f}")   # Unterstrich als Tausendertrennzeichen
````

💡 Das Format folgt der englischen Schreibweise mit Komma als Tausender- und Punkt als Dezimaltrennzeichen. Für eine deutsche Darstellung lassen sich die Zeichen anschließend tauschen, etwa mit `.replace()`.

Mit einer Breite und einem Ausrichtungszeichen (`<` links, `>` rechts, `^` zentriert) entstehen bündige Tabellen:

````python
products = [("Kaffee", 4.99), ("Brot", 2.49), ("Olivenöl", 12.5)]

for product, price in products:
    print(f"{product:<10} {price:>8.2f} Euro")
````

## 5. Verkettete Vergleiche

Prüfungen, ob ein Wert in einem Bereich liegt, schreibt man bisher mit `and`. Python erlaubt stattdessen, Vergleiche wie in der Mathematik zu verketten:

````python
temperature = 22

# Mit and
if temperature >= 18 and temperature <= 25:
    print("Angenehme Temperatur (mit and)")

# Verkettet
if 18 <= temperature <= 25:
    print("Angenehme Temperatur (verkettet)")
````

`18 <= temperature <= 25` bedeutet dasselbe wie `18 <= temperature and temperature <= 25`. Verkettungen funktionieren mit allen Vergleichsoperatoren, sind aber vor allem für Bereichsprüfungen wie diese gut lesbar.

## 6. Das `with`-Statement und Context Manager

Aus der Session "Dateien" ist das `with`-Statement bekannt: Es öffnet eine Datei und schließt sie nach dem eingerückten Block automatisch wieder, auch wenn im Block ein Fehler auftritt.

````python
with open("beispiel.txt", "w", encoding="utf-8") as example_file:
    example_file.write("Hallo mit with-Statement!\n")

# Hier ist die Datei bereits wieder geschlossen
print(f"{example_file.closed = }")
````

Hinter `with` steht ein allgemeines Prinzip: Ein **Context Manager** ist ein Objekt, das festlegt, was beim **Betreten** und beim **Verlassen** eines Blocks passieren soll. Bei Dateien heißt das: beim Betreten öffnen, beim Verlassen schließen. Dasselbe Prinzip nutzen zum Beispiel Datenbankverbindungen, die nach Gebrauch wieder freigegeben werden müssen.

Ein `with`-Statement kann mehrere Context Manager gleichzeitig verwalten, etwa um aus einer Datei zu lesen und in eine andere zu schreiben:

````python
with open("quelle.txt", "w", encoding="utf-8") as source_file:
    source_file.write("Quelltext")

with open("quelle.txt", "r", encoding="utf-8") as source_file, open("ziel.txt", "w", encoding="utf-8") as target_file:
    text = source_file.read()
    target_file.write("Bearbeiteter " + text)

with open("ziel.txt", "r", encoding="utf-8") as target_file:
    print(target_file.read())
````

💡 Eigene Context Manager werden mit Klassen und den Methoden `__enter__` und `__exit__` geschrieben. Das ist Thema der Advanced Sessions zur objektorientierten Programmierung.

## 7. Type Hints

Mit **Type Hints** wird im Code angegeben, welche Datentypen ein Parameter erwartet und welchen Typ eine Funktion zurückgibt. Ähnlich wie die Beschriftung an einem Stecker zeigen sie auf einen Blick, was wo hineingehört.

````python
def greet(name: str, repeat: int = 1) -> str:
    """Return a greeting for name, repeated repeat times."""
    return f"Hallo {name}! " * repeat


print(greet("Anna"))
print(greet("Bernd", 2))
````

- `name: str`: Der Parameter `name` erwartet einen String.
- `repeat: int = 1`: Type Hint und Standardwert lassen sich kombinieren.
- `-> str`: Die Funktion gibt einen String zurück.

⚠️ Python **prüft** Type Hints beim Ausführen nicht. Ein falscher Typ führt nicht zu einem Fehler durch den Type Hint, sondern höchstens zu einem Fehler im Code selbst:

````python
print(greet(42))   # läuft trotzdem, obwohl 42 kein String ist
````

Type Hints sind trotzdem nützlich: Sie dokumentieren die Funktion, Editoren wie VS Code zeigen sie beim Aufruf an und warnen bei offensichtlichen Fehlern, und Werkzeuge wie `mypy` können sie automatisch prüfen.

Für Datenstrukturen wird der Typ der Elemente in eckigen Klammern angegeben. Mit `|` lassen sich mehrere erlaubte Typen angeben, zum Beispiel für Werte, die auch `None` sein dürfen:

````python
def average(values: list[float]) -> float | None:
    """Return the average of values, or None for an empty list."""
    if not values:
        return None
    return sum(values) / len(values)


def count_words(text: str) -> dict[str, int]:
    """Count how often each word occurs in text."""
    counts = {}
    for word in text.lower().split():
        counts[word] = counts.get(word, 0) + 1
    return counts


print(average([2.5, 3.5, 4.0]))
print(average([]))
print(count_words("Die Katze ist eine Katze"))
````

💡 In älterem Code finden sich oft Schreibweisen wie `List[int]`, `Dict[str, int]` oder `Optional[str]` aus dem Modul `typing`. Seit Python 3.9 bzw. 3.10 sind `list[int]`, `dict[str, int]` und `str | None` die übliche Schreibweise.

## 8. Zusammenfassung

Die Konstrukte dieser Session machen häufige Muster kürzer. Sie ersetzen keine bisher bekannten Werkzeuge, sondern bieten eine kompaktere Schreibweise, die eingesetzt werden sollte, wenn sie den Code verständlicher macht.

### Behandelte Konzepte

1. Mit `a if condition else b` zwischen zwei Werten wählen und bei mehr Fällen bei `if`-`elif`-`else` bleiben.
2. Mit List-, Dictionary- und Set-Comprehensions Datenstrukturen erzeugen, mit `if` am Ende filtern und mit einem ternären Operator vorne umformen.
3. Werte mit Tuple Unpacking verteilen, mit `*` restliche Elemente sammeln und mit `*` bzw. `**` Argumente beim Aufruf entpacken.
4. Zahlen in f-Strings mit Formatangaben wie `:.2f`, `:,` oder `:>8` darstellen.
5. Bereichsprüfungen als verkettete Vergleiche wie `18 <= value <= 25` schreiben.
6. Das `with`-Statement als Context Manager verstehen, der Ressourcen beim Verlassen des Blocks freigibt.
7. Parameter und Rückgabewerte mit Type Hints dokumentieren und wissen, dass Python sie nicht prüft.

### Kurzreferenz

| Schreibweise | Bedeutung |
|---|---|
| `a if condition else b` | `a`, wenn die Bedingung zutrifft, sonst `b` |
| `[expr for x in items]` | neue Liste aus jedem Element |
| `[expr for x in items if condition]` | neue Liste, gefiltert |
| `[a if condition else b for x in items]` | neue Liste, je Element umgeformt |
| `{key: value for x in items}` | neues Dictionary |
| `{expr for x in items}` | neues Set |
| `first, second = pair` | Tuple Unpacking |
| `first, *rest = items` | restliche Elemente als Liste sammeln |
| `func(*my_list)`, `func(**my_dict)` | Liste bzw. Dictionary als Argumente übergeben |
| `{**dict_a, **dict_b}` | Dictionaries zusammenführen |
| `f"{value:.2f}"` | 2 Nachkommastellen |
| `f"{value:,.2f}"` | mit Tausendertrennzeichen |
| `f"{text:<10}"`, `f"{value:>8}"` | links- bzw. rechtsbündig mit fester Breite |
| `low <= value <= high` | verketteter Vergleich |
| `with open(...) as file:` | Context Manager |
| `def func(name: str) -> int:` | Type Hints |
| `list[int]`, `dict[str, int]`, `str \| None` | Type Hints für Datenstrukturen und optionale Werte |

### Ausblick

In der Session "Code-Qualität und Refactoring" wird bestehender Code Schritt für Schritt verbessert. Comprehensions, f-Strings und Type Hints sind dort Werkzeuge, um Code lesbarer zu machen.

### 9.1 Generator Expressions

Mit runden statt eckigen Klammern entsteht keine Liste, sondern eine **Generator Expression**. Sie berechnet die Werte erst, wenn sie gebraucht werden, und hält nie alle gleichzeitig im Speicher:

````python
squares_list = [number**2 for number in range(10)]
squares_generator = (number**2 for number in range(10))

print(squares_list)
print(squares_generator)   # zeigt nur das Generator-Objekt, noch keine Werte
````

Generator Expressions eignen sich vor allem, wenn die Werte direkt weiterverarbeitet werden, zum Beispiel von `sum()`, `max()` oder `tuple()`. Steht die Generator Expression als einziges Argument in einem Funktionsaufruf, können die eigenen Klammern entfallen:

````python
print(sum(number**2 for number in range(10)))
print(tuple(number**2 for number in range(5)))   # eine "Tuple Comprehension" gibt es nicht
````

<a id="nested-comprehensions"></a>

### 9.2 Verschachtelte Comprehensions

Comprehensions lassen sich verschachteln, etwa für Tabellen aus Zeilen und Spalten:

````python
multiplication_table = [[row * column for column in range(1, 6)] for row in range(1, 6)]

for table_row in multiplication_table:
    print(table_row)
````

Zwei `for`-Teile hintereinander in einer einzigen Comprehension machen aus einer verschachtelten Liste eine flache. Die Reihenfolge der `for`-Teile entspricht dabei der Reihenfolge verschachtelter Schleifen:

````python
flat_values = [value for table_row in multiplication_table for value in table_row]
print(flat_values)
````

<a id="walrus"></a>

### 9.3 Der Walrus-Operator `:=`

Der Walrus-Operator `:=` (benannt nach seiner Ähnlichkeit mit einem liegenden Walross) weist einen Wert zu **und** gibt ihn gleichzeitig zurück. Damit lässt sich ein Zwischenergebnis innerhalb einer Bedingung speichern:

````python
text = "Hallo Welt"

# Ohne Walrus-Operator
length = len(text)
if length > 5:
    print(f"'{text}' hat {length} Zeichen.")

# Mit Walrus-Operator
if (length := len(text)) > 5:
    print(f"'{text}' hat {length} Zeichen.")
````

In Comprehensions vermeidet der Walrus-Operator, dass ein Ausdruck für Bedingung und Ergebnis doppelt berechnet wird:

````python
numbers = [1, 2, 3, 4, 5]

large_squares = [number**2 for number in numbers if number**2 > 10]           # zweimal berechnet
large_squares = [square for number in numbers if (square := number**2) > 10]   # einmal berechnet
print(large_squares)
````

### 9.4 Structural Pattern Matching mit `match`-`case`

Seit Python 3.10 lässt sich ein Wert mit `match`-`case` gegen mehrere Muster prüfen. Für einfache Fälle ähnelt das einer `if`-`elif`-Kette:

````python
def describe_status(status_code: int) -> str:
    """Return a description for an HTTP status code."""
    match status_code:
        case 200:
            return "OK"
        case 404:
            return "Not Found"
        case 500 | 503:            # mehrere Werte mit |
            return "Serverfehler"
        case _:                    # alle anderen Fälle
            return f"Unbekannter Status: {status_code}"


for code_to_check in [200, 404, 503, 302]:
    print(f"{code_to_check}: {describe_status(code_to_check)}")
````

Seine Stärken zeigt `match`-`case` bei strukturierten Daten: Muster können Tupel oder Listen zerlegen, die Bestandteile direkt Variablen zuweisen und mit `if` zusätzliche Bedingungen (*Guards*) prüfen:

````python
def describe_point(point: tuple[int, int]) -> str:
    """Describe where a point lies in the coordinate system."""
    match point:
        case (0, 0):
            return "Ursprung"
        case (0, y):
            return f"auf der y-Achse bei y={y}"
        case (x, 0):
            return f"auf der x-Achse bei x={x}"
        case (x, y) if x == y:
            return f"auf der Diagonalen bei ({x}, {y})"
        case (x, y):
            return f"bei ({x}, {y})"


for point in [(0, 0), (0, 5), (3, 0), (2, 2), (3, 4)]:
    print(f"{point}: {describe_point(point)}")
````

<a id="precedence"></a>

### 9.5 Operatorreihenfolge

Python wertet Operatoren in einer festen Reihenfolge aus, ähnlich wie "Punkt vor Strich" in der Mathematik. Vereinfacht, von höchster zu niedrigster Priorität:

1. Klammern `()`
2. Potenz `**`
3. Vorzeichen `-x`
4. `*`, `/`, `//`, `%`
5. `+`, `-`
6. Vergleiche `<`, `<=`, `>`, `>=`, `!=`, `==`, `is`, `in`
7. `not`
8. `and`
9. `or`

````python
first_value, second_value, third_value = 5, 10, 15

# Rechnen vor Vergleichen: (5 + 10) < 15
print(f"{first_value + second_value < third_value = }")

# not vor and vor or
print(f"{not first_value < second_value and third_value > second_value or first_value == 5 = }")

# Dasselbe mit Klammern, deutlich lesbarer
print(f"{((not first_value < second_value) and third_value > second_value) or first_value == 5 = }")
````

💡 Wer die Reihenfolge nachschlagen muss, sollte Klammern setzen. Sie ändern am Ergebnis nichts, machen die Absicht aber sofort sichtbar.

<a id="old-formatting"></a>

### 9.6 Ältere Formen der String-Formatierung

f-Strings gibt es seit Python 3.6. In älterem Code und in vielen Online-Beispielen finden sich noch zwei ältere Schreibweisen, die zum selben Ergebnis führen:

````python
name = "Anna"
age = 30

print("%s ist %d Jahre alt." % (name, age))          # %-Formatierung
print("{} ist {} Jahre alt.".format(name, age))      # str.format()
print(f"{name} ist {age} Jahre alt.")                # f-String
````

### Type Hints

#### Aufgabe 14: Funktionen annotieren

Ergänze bei beiden Funktionen passende Type Hints für alle Parameter und den Rückgabewert. `find_product` gibt `None` zurück, wenn das Produkt nicht gefunden wird.

Rufe `apply_discount` anschließend mit einem String statt einer Zahl auf. Was passiert, und warum hilft der Type Hint hier nicht?

````python
def apply_discount(price, discount_percent=10):
    """Return the price reduced by discount_percent."""
    return round(price * (1 - discount_percent / 100), 2)


def find_product(products, name):
    """Return the price of the product with the given name, or None."""
    return products.get(name)


print(apply_discount(19.99))                              # erwartet: 17.99
print(find_product({"Kaffee": 4.99}, "Kaffee"))           # erwartet: 4.99
print(find_product({"Kaffee": 4.99}, "Tee"))              # erwartet: None
````
